# Python-Based Analysis of Cooling Curves in Medium-Carbon Steel

#### Project Information:

Material: Medium-carbon steel (~0.4 wt% C)
Initial temperature: 850°C
Cooling conditions: Furnace cooling, air cooling, and water cooling
Data type: Simulated cooling-temperature data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#time data
time = np.arange(0, 601, 10)

print(time)
```

### Furnace Cooling

Furnace cooling is a slow cooling process. To show the gradual decrease in temperature,we use:


### T(t)=(850-25)e^{-0.0025t}+25


Here, 850°C is the initial temperature and 25°C is room temperature.

```python
#furnace temp.
furnace_temp = (850 - 25) * np.exp(-0.0025 * time) + 25

print(furnace_temp)
```

```python
#now showing graphically
plt.plot(time,furnace_temp)

#labeling x and y
plt.xlabel("Time (sec)")
plt.ylabel("Temperature (°C)")

plt.title("Furnace Cooling Curve") #one title giving for graph
plt.show() #--->showing graph
```

### Air Cooling

Air cooling is faster than furnace cooling.
    general eq:  T(t) = (850 - 25)e^(-kt) + 25,

for air cooling k=0.006 (assume)

```python
air_temp = (850 - 25) * np.exp(-0.006 * time) + 25

print(air_temp)
```

```python
#now showing graphically
plt.plot(time,air_temp)

#labeling x and y
plt.xlabel("Time (sec)")
plt.ylabel("Temperature (°C)")

plt.title("Air Cooling curve") #one title giving for graph
plt.show() #--->showing graph
```

```python
#comparing both graphs
plt.plot(time,furnace_temp,label="furnace cooling")

plt.plot(time,air_temp,label="air cooling")

#labeling x and y
plt.xlabel("Time (sec)")
plt.ylabel("Temperature (°C)")

plt.title("Comparison of both furnace and air cooling") #one title giving for graph

plt.legend() # for identify curve
plt.show()
```

### Water Cooling

Water cooling is a fastest cooling process.
T(t) = (850 - 25)e^(-kt) + 25,

for air cooling k=0.02 (assume)

```python
water_temp = (850 - 25) * np.exp(-0.02 * time) + 25

print(water_temp)
```

### Now ploting graph for all in one to compare

```python
plt.plot(time, furnace_temp, label="Furnace Cooling")
plt.plot(time, air_temp, label="Air Cooling")
plt.plot(time, water_temp, label="Water Cooling")

#labeling x and y
plt.xlabel("Time (s)")
plt.ylabel("Temperature (°C)")

plt.title("Cooling Curves for Different Cooling Conditions")
plt.legend()
plt.show()
```

```python
data = pd.DataFrame({
    "Time(s)": time,
    "Furnace.Temp": furnace_temp,
    "Air.Temp": air_temp,
    "Water.Temp": water_temp
})

print("Data (temp. are in celsius) :\n")
print("\n",data)
```

```python
#convert data into csv
data.to_csv("../data/cooling_data.csv", index=False)
```

### Cooling Rate ; CoolingRate= -dT/dt

```python
#for furnace rate

furnace_rate = -np.gradient(furnace_temp, time)
print(furnace_rate)
```

```python
#for air rate

air_rate = -np.gradient(air_temp, time)
print(air_rate)
```

```python
#for Water rate

water_rate = -np.gradient(water_temp, time)
print(water_rate)
```

### maximum cooling rate for all condtion

```python
#For Furnace,Air,Water

#to find maximum
x = np.max(furnace_rate)
y= np.max(air_rate)
z= np.max(water_rate)
print("Furnace:",x, "°C/s")
print("Air:",y,"°C/s")
print("Water:",z, "°C/s")
```

Furnace < Air < Water
Means water cooling is the fastest, while furnace cooling is the slowest.

### Cooling rate vs time graph

```python
plt.plot(time, furnace_rate, label="Furnace Cooling")
plt.plot(time, air_rate, label="Air Cooling")
plt.plot(time, water_rate, label="Water Cooling")

#labeling x and y
plt.xlabel("Time (s)")
plt.ylabel("Cooling Rate (°C/s)")

plt.title("Cooling Rate vs Time")
plt.legend()
plt.show()
```

##### The cooling rate of water is highest compare to furnace and air cooling rate.

## Metallurical Interpretation
Cooling rate affects how the steel changes the microstructure and mechanical properties of Steel.

### Effect of Cooling Conditions

- Furnace cooling: Slow cooling → generally coarser structure and lower hardness.
- Air cooling: Moderate cooling → finer structure and moderate hardness.
- Water cooling: Fast cooling → may form a harder structure.

## Cooling Rate Comparison

- Furnace cooling: 2.04 °C/s
- Air cooling: 4.80 °C/s
- Water cooling: 14.95 °C/s

Water cooling has the highest cooling rate and furnace cooling has the lowest.

## Conclusion

The simulated results show that cooling rate depend on the cooling condition. Furnace cooling is the slowest, air cooling is faster, and water cooling is the fastest.

The results also show that cooling rate can affect the microstructure and mechanical properties of medium-carbon steel.
